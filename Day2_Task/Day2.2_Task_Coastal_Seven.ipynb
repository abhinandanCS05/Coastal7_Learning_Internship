# DAY 2  – Data Structures + File Handling



## 1. Lists

 A list stores multiple values. It is ordered, changeable, and allows duplicates.

```python
students = ["Abhi", "Rahul", "Priya"]
print(students)
```

## 2. List indexing and slicing

 Indexing accesses one item. Slicing accesses a part of a list. Python indexing starts at 0.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[-1])
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
```

## 3. List CRUD

 CRUD means Create, Read, Update, and Delete.

```python
students = ["Abhi", "Rahul"]

# Create
students.append("Priya")

# Read
print(students)

# Update
students[0] = "Abhinandan"

# Delete
students.remove("Rahul")

print(students)
```

## 4. sort() and reverse()

sort() arranges list items. reverse() reverses the order.

```python
numbers = [5, 2, 8, 1, 3]

numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)
```

## 5. Tuples and immutability

 A tuple is ordered but cannot be changed after creation. This is called immutability.

```python
student = ("Abhi", 21, "AIML")

print(student)
print(student[0])

# This would cause an error:
# student[1] = 22
```

## 6. Sets

 A set stores unique, unordered values.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print("Union:", a.union(b))
print("Intersection:", a.intersection(b))
print("Difference:", a.difference(b))
```

## 7. Dictionaries

 A dictionary stores data as key-value pairs.

```python
student = {
    "name": "Abhi",
    "age": 21,
    "branch": "AIML"
}

print(student)
print(student.get("name"))
print(student.get("college", "Not available"))
print(student.keys())
print(student.items())
```

## 8. Dictionary update, delete and comprehension

 Dictionary values can be changed, removed, or generated using comprehension.

```python
student = {"name": "Abhi", "age": 21}

student["age"] = 22
student["branch"] = "AIML"
del student["age"]

print(student)

numbers = [1, 2, 3, 4]
squares = {number: number ** 2 for number in numbers}

print(squares)
```

## 9. Nested lists and dictionaries

 A nested structure contains one list or dictionary inside another.

```python
marks = [[80, 85, 90], [70, 75, 78]]

print(marks[0][1])

students = {
    "student1": {"name": "Abhi", "branch": "AIML"},
    "student2": {"name": "Rahul", "branch": "CSE"}
}

print(students["student1"]["name"])
print(students["student2"]["branch"])
```

## 10. JSON write

 JSON is a format used to store structured data. `json.dump()` writes Python data to a JSON file.

```python
import json

student = {"name": "Abhi", "age": 21, "branch": "AIML"}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

print("JSON file created")
```

## 11. JSON read

`json.load()` reads JSON data from a file.

```python
import json

with open("student.json", "r") as file:
    student_data = json.load(file)

print(student_data)
```

## 12. CSV write

 CSV means Comma-Separated Values. It is useful for table-like data.

```python
import csv

students = [
    ["Name", "Age", "Branch"],
    ["Abhi", 21, "AIML"],
    ["Rahul", 22, "CSE"]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(students)

print("CSV file created")
```

## 13. CSV read

 `csv.reader()` reads rows from a CSV file.

```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

## 14. Basic try/except

 `try/except` handles runtime errors safely.

```python
try:
    number = int(input("Enter a number: "))
    print(10 / number)
except ValueError:
    print("Enter a valid number")
except ZeroDivisionError:
    print("Cannot divide by zero")
```

## 15. Contact Book – JSON persistence

Persistence means saving data in a file so it remains available after the program ends.

```python
import json

def save_contacts(contacts):
    with open("contacts.json", "w") as file:
        json.dump(contacts, file, indent=4)

def load_contacts():
    try:
        with open("contacts.json", "r") as file:
            return json.load(file)
    except FileNotFoundError:
        return {}

def add_contact(contacts, name, phone):
    contacts[name] = phone
    save_contacts(contacts)

def view_contacts(contacts):
    if not contacts:
        print("No contacts found")
    else:
        for name, phone in contacts.items():
            print(name, ":", phone)

def search_contact(contacts, name):
    phone = contacts.get(name)
    if phone:
        print(name, ":", phone)
    else:
        print("Contact not found")

def delete_contact(contacts, name):
    if name in contacts:
        del contacts[name]
        save_contacts(contacts)
        print("Contact deleted")
    else:
        print("Contact not found")

contacts = load_contacts()

add_contact(contacts, "Abhi", "9876543210")
add_contact(contacts, "Rahul", "9876501234")

view_contacts(contacts)
search_contact(contacts, "Abhi")
delete_contact(contacts, "Rahul")
view_contacts(contacts)
```
