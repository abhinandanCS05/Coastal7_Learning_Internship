#  DAY 2  – Python Control Flow & Functions


## 1. if / elif / else

 Used to execute different blocks of code based on a condition.

```python
age = 21

if age >= 18:
    print("Adult")
elif age > 0:
    print("Minor")
else:
    print("Invalid age")
```

## 2. for loop

 A for loop repeats a block of code for each item in a sequence.

```python
for number in range(1, 6):
    print(number)
```

## 3. while loop

 A while loop repeats code as long as a condition is true.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

## 4. break

 break stops the loop immediately.

```python
for number in range(1, 6):
    if number == 3:
        break
    print(number)
```

## 5. continue

continue skips the current iteration and moves to the next iteration.

```python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

## 6. pass
pass does nothing. It is used as a placeholder when code is required syntactically.

```python
for number in range(3):
    if number == 1:
        pass
    print(number)
```

## 7. range()

range() generates a sequence of numbers. It is commonly used with loops.

```python
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(2, 11, 2)))
```

## 8. List comprehension

 A list comprehension is a short way to create a list using a loop.

```python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
```

## 9. Functions

 A function is a reusable block of code that performs a specific task.

```python
def greet():
    print("Hello, Abhi!")

greet()
```

## 10. Function parameters

 Parameters allow us to pass data into a function.

```python
def greet_user(name):
    print("Hello", name)

greet_user("Abhi")
```

## 11. return statement
 return sends a result back from a function.

```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

## 12. Default arguments

 A default argument is used when the caller does not provide a value.

```python
def welcome(name, message="Welcome"):
    print(message, name)

welcome("Abhi")
welcome("Abhi", "Good morning")
```

## 13. *args

args allows a function to accept multiple positional arguments.

```python
def add_all(*numbers):
    total = 0

    for number in numbers:
        total += number

    return total

print(add_all(1, 2, 3))
print(add_all(10, 20))
```

## 14. **kwargs

 **kwargs allows a function to accept multiple keyword arguments.

```python
def show_details(**details):
    for key, value in details.items():
        print(key, ":", value)

show_details(name="Abhi", branch="AIML")
```

## 15. Returning multiple values
 A function can return more than one value as a tuple.

```python
def calculate(a, b):
    return a + b, a - b

addition, subtraction = calculate(10, 5)

print(addition)
print(subtraction)
```

## 16. Lambda function

 A lambda is a small anonymous function written in one line.

```python
square = lambda number: number ** 2

print(square(5))
```

## 17. Local scope

 A local variable is created inside a function and is normally available only inside that function.

```python
def show_name():
    name = "Abhi"
    print(name)

show_name()
```

## 18. Global scope

 A global variable is created outside functions and can be accessed from inside functions.

```python
name = "Abhi"

def display_name():
    print(name)

display_name()
```

## 19. LEGB rule

 Python searches for variables in this order: Local, Enclosing, Global, Built-in.

```python
name = "Global"

def outer():
    name = "Enclosing"

    def inner():
        name = "Local"
        print(name)

    inner()

outer()
print(name)
```

## 20. Practice – FizzBuzz

 Print Fizz for multiples of 3, Buzz for multiples of 5, and `FizzBuzz` for multiples of both.

```python
def fizz_buzz(limit):
    for number in range(1, limit + 1):
        if number % 3 == 0 and number % 5 == 0:
            print("FizzBuzz")
        elif number % 3 == 0:
            print("Fizz")
        elif number % 5 == 0:
            print("Buzz")
        else:
            print(number)

fizz_buzz(15)
```

## 21. Practice – Prime number checker

 A prime number is a number greater than 1 that has only two factors: 1 and itself.

```python
def is_prime(number):
    if number <= 1:
        return False

    for divisor in range(2, number):
        if number % divisor == 0:
            return False

    return True

print(is_prime(7))
print(is_prime(10))
```

## 22. Practice – Temperature converter

 Temperature conversion changes a temperature from Celsius to Fahrenheit or Fahrenheit to Celsius.

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

print(celsius_to_fahrenheit(25))
print(fahrenheit_to_celsius(77))
```

## 23. Final mini program – Number checker

 This program uses a function and conditions to identify whether a number is positive, negative, or zero.

```python
def check_number(number):
    if number > 0:
        return "Positive"
    elif number < 0:
        return "Negative"
    else:
        return "Zero"

print(check_number(10))
print(check_number(-5))
print(check_number(0))
```
