# **EXCEPTION HANDLING**

 Used to handle errors without crashing the program.

```python

try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

    print("Result:", result)

except ValueError:
    print("Please enter valid numbers.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Program completed.")
```

# **SIMPLE PRACTICAL PROGRAM**



##  Student **GRADE CALCULATOR** using dictionary, if/else, function and exception handling.

```python
def student_grade():
    try:
        name = input("Enter student name: ")
        marks = float(input("Enter marks: "))

        if marks < 0 or marks > 100:
            print("Marks must be between 0 and 100.")
            return

        if marks >= 90:
            grade = "A+"
        elif marks >= 75:
            grade = "A"
        elif marks >= 60:
            grade = "B"
        else:
            grade = "C"

        student = {
            "name": name,
            "marks": marks,
            "grade": grade
        }

        print("\nStudent Result:")
        for key, value in student.items():
            print(key, ":", value)

    except ValueError:
        print("Invalid input. Enter a number for marks.")


student_grade()
```

```python
from fastapi import FastAPI
from pydantic import BaseModel
import nest_asyncio
import uvicorn

# Allow server to run in Colab
nest_asyncio.apply()

# Create FastAPI app
app = FastAPI()


# Pydantic model
class Student(BaseModel):
    name: str
    age: int
    branch: str


# Simple database
students = {}


# GET - Get all students
@app.get("/students")
def get_students():
    return students


# GET - Get student by ID
@app.get("/students/{student_id}")
def get_student(student_id: int):
    return students.get(student_id, "Student not found")


# GET - Query parameter
@app.get("/search")
def search_student(branch: str):
    return {
        "branch": branch,
        "message": "Searching students"
    }


# POST - Add student
@app.post("/students/{student_id}")
def add_student(student_id: int, student: Student):
    students[student_id] = student.model_dump()

    return {
        "message": "Student added",
        "student": students[student_id]
    }


# PUT - Update student
@app.put("/students/{student_id}")
def update_student(student_id: int, student: Student):
    students[student_id] = student.model_dump()

    return {
        "message": "Student updated",
        "student": students[student_id]
    }


# DELETE - Delete student
@app.delete("/students/{student_id}")
def delete_student(student_id: int):
    student = students.pop(student_id, None)

    return {
        "message": "Student deleted",
        "student": student
    }


# Run server

import threading

def run_server():
    uvicorn.run(
        app,
        host="0.0.0.0",
        port=8000
    )

server_thread = threading.Thread(
    target=run_server,
    daemon=True
)

server_thread.start()

print("FastAPI server started!")
print("Running on port 8000")
```
